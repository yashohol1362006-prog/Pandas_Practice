## ### Merging Joining Concatiation

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
print(pd.merge(employees,salaries, on='employee_id', how='outer')) 
print(pd.merge(employees, salaries,on='employee_id',how='left'))   # if we merge 2 dataframes on the basis of employees_id and how=left then from the the 2 data framae it'll will consider 1st on 
print(pd. merge(employees, salaries, on='employee_id', how='right' ))
print(employees)
print(salaries)
print(pd.merge(employees, salaries, on='employee_id', how='inner'))  #in inner joint common data gets combine
print(pd.merge(employees, salaries,))
print(pd.merge(employees, salaries, on='employee_id',how='outer' ))  #in outer joint non-commmon data gets combuined and blank data is filled with Nan
print(pd.merge(employees, salaries, on= 'employee_id', how='left')) # it will print employee id on the basis of employees(which is left thats whyh we have used left)
print(pd.merge(employees, salaries, on='employee_id', how='right')) # it will print employee id on the basis of salaries (which is right thats why we have used right)


## concatanation 
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
print(pd.concat([df1, df2]))
print(pd.concat([df2, df1]))
print(pd.concat([df1, df2], axis=1))
print(pd.concat([df2, df1], axis=1))
```

## Pivot Table ####

```python

import pandas as pd
import pandas as pd

data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South',
               'East', 'West', 'North', 'South',
               'East', 'West', 'North', 'South',
               'East', 'West', 'North', 'South',
               'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice',
            'John', 'Mary', 'Bob', 'Alice',
            'John', 'Mary', 'Bob', 'Alice',
            'John', 'Mary', 'Bob', 'Alice',
            'John', 'Mary', 'Bob', 'Alice']
}
df = pd.DataFrame(data)
df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)
print(df)
pivot  = df.pivot_table(values=['Sales','Units'],
                        index='Region',
                        columns='Rep',
                        aggfunc= 'max')
print(pivot)
```
